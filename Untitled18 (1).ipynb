Ridership by Day and Hour

```python
# Step 4.1 - Create copy for temporal analysis

time_df = df.copy()

print("Temporal analysis dataset created.")
print("Number of records:", len(time_df))
```

```python
# Step 4.2 - Convert pickup_datetime to datetime

time_df['pickup_datetime'] = pd.to_datetime(
    time_df['pickup_datetime'],
    errors='coerce'
)

print("Data type:")
print(time_df['pickup_datetime'].dtype)
```

```python
print("Missing pickup_datetime values:",
      time_df['pickup_datetime'].isna().sum())
```

```python
# Step 4.3 - Extract hour

time_df['hour'] = time_df['pickup_datetime'].dt.hour

print("Hour extraction completed.")

display(
    time_df[['pickup_datetime', 'hour']].head(10)
)
```

```python
# Step 4.4 - Extract day of month

time_df['day'] = time_df['pickup_datetime'].dt.day

display(
    time_df[['pickup_datetime', 'day']].head(10)
)
```

```python
# Step 4.5 - Extract day of week

time_df['day_of_week'] = time_df['pickup_datetime'].dt.dayofweek

time_df['day_name'] = time_df['pickup_datetime'].dt.day_name()

display(
    time_df[
        ['pickup_datetime', 'day_of_week', 'day_name']
    ].head(10)
)
```

```python
# Step 4.6 - Extract month

time_df['month'] = time_df['pickup_datetime'].dt.month

time_df['month_name'] = time_df['pickup_datetime'].dt.month_name()

display(
    time_df[
        ['pickup_datetime', 'month', 'month_name']
    ].head(10)
)
```

```python
# Step 4.7 - Extract year

time_df['year'] = time_df['pickup_datetime'].dt.year

display(
    time_df[
        ['pickup_datetime', 'year']
    ].head(10)
)
```

```python
print("Years present in dataset:")

display(
    time_df['year'].value_counts().sort_index()
)
```

```python
# Step 4.8 - Weekday vs Weekend

time_df['day_type'] = np.where(
    time_df['day_of_week'] < 5,
    'Weekday',
    'Weekend'
)

display(
    time_df[
        ['day_name', 'day_of_week', 'day_type']
    ].head(10)
)
```

```python
# Step 4.9 - Trips by hour

trips_by_hour = time_df['hour'].value_counts().sort_index()

display(trips_by_hour)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=trips_by_hour.index,
    y=trips_by_hour.values
)

plt.title("Number of Taxi Trips by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Trips")

plt.xticks(range(24))

plt.show()
```

```python
# Step 4.10 - Peak and off-peak hours

peak_hour = trips_by_hour.idxmax()
peak_trips = trips_by_hour.max()

off_peak_hour = trips_by_hour.idxmin()
off_peak_trips = trips_by_hour.min()

print("Peak hour:", peak_hour)
print("Trips during peak hour:", peak_trips)

print("\nOff-peak hour:", off_peak_hour)
print("Trips during off-peak hour:", off_peak_trips)
```

```python
# Step 4.11 - Trips by day of week

day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

trips_by_day = (
    time_df['day_name']
    .value_counts()
    .reindex(day_order)
)

display(trips_by_day)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=trips_by_day.index,
    y=trips_by_day.values
)

plt.title("Number of Taxi Trips by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Trips")

plt.xticks(rotation=30)

plt.show()
```

```python
# Step 4.12 - Highest and lowest ridership days

highest_day = trips_by_day.idxmax()
highest_day_trips = trips_by_day.max()

lowest_day = trips_by_day.idxmin()
lowest_day_trips = trips_by_day.min()

print("Highest ridership day:", highest_day)
print("Number of trips:", highest_day_trips)

print("\nLowest ridership day:", lowest_day)
print("Number of trips:", lowest_day_trips)
```

```python
# Step 4.13 - Trips by month

trips_by_month = (
    time_df['month']
    .value_counts()
    .sort_index()
)

display(trips_by_month)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=trips_by_month.index,
    y=trips_by_month.values
)

plt.title("Number of Taxi Trips by Month")
plt.xlabel("Month")
plt.ylabel("Number of Trips")

plt.xticks(range(12), [
    'Jan', 'Feb', 'Mar', 'Apr',
    'May', 'Jun', 'Jul', 'Aug',
    'Sep', 'Oct', 'Nov', 'Dec'
])

plt.show()
```

```python
# Step 4.14 - Trips by year

trips_by_year = (
    time_df['year']
    .value_counts()
    .sort_index()
)

display(trips_by_year)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=trips_by_year.index.astype(str),
    y=trips_by_year.values
)

plt.title("Number of Taxi Trips by Year")
plt.xlabel("Year")
plt.ylabel("Number of Trips")

plt.show()
```

```python
# Step 4.15 - Weekday vs Weekend

trips_by_day_type = time_df['day_type'].value_counts()

display(trips_by_day_type)
```

```python
plt.figure(figsize=(8, 6))

sns.barplot(
    x=trips_by_day_type.index,
    y=trips_by_day_type.values
)

plt.title("Weekday vs Weekend Taxi Ridership")
plt.xlabel("Day Type")
plt.ylabel("Number of Trips")

plt.show()
```

```python
# Step 4.16 - Day of week x hour

day_hour = pd.crosstab(
    time_df['day_name'],
    time_df['hour']
)

day_hour = day_hour.reindex(day_order)

display(day_hour)
```

```python
plt.figure(figsize=(16, 7))

sns.heatmap(
    day_hour,
    cmap='YlOrRd',
    annot=False
)

plt.title("Taxi Trips by Day of Week and Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Day of Week")

plt.show()
```

```python
# Step 4.17 - Busiest day and hour combination

max_position = day_hour.stack().idxmax()
max_trips = day_hour.stack().max()

print("Busiest day:", max_position[0])
print("Busiest hour:", max_position[1])
print("Number of trips:", max_trips)
```

```python
min_position = day_hour.stack().idxmin()
min_trips = day_hour.stack().min()

print("Least busy day:", min_position[0])
print("Least busy hour:", min_position[1])
print("Number of trips:", min_trips)
```

```python
# Step 4.18 - Average fare by hour

average_fare_by_hour = (
    time_df
    .groupby('hour')['fare_amount']
    .mean()
)

display(average_fare_by_hour)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    x=average_fare_by_hour.index,
    y=average_fare_by_hour.values,
    marker='o'
)

plt.title("Average Taxi Fare by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Fare")

plt.xticks(range(24))

plt.show()
```

```python
# Step 4.19 - Average fare: weekday vs weekend

average_fare_day_type = (
    time_df
    .groupby('day_type')['fare_amount']
    .mean()
)

display(average_fare_day_type)
```

```python
plt.figure(figsize=(8, 6))

sns.barplot(
    x=average_fare_day_type.index,
    y=average_fare_day_type.values
)

plt.title("Average Taxi Fare: Weekday vs Weekend")
plt.xlabel("Day Type")
plt.ylabel("Average Fare")

plt.show()
```

```python
# ============================================================
# STEP 5.12 — DEFINE CLEANING RULES
# ============================================================

print("=" * 70)
print("STEP 5.12 — DEFINING CLEANING RULES")
print("=" * 70)

# 1. Invalid fare
invalid_fare = df['fare_amount'] <= 0

# 2. Invalid passenger count
invalid_passenger = (
    (df['passenger_count'] <= 0) |
    (df['passenger_count'] > 6)
)

# 3. Missing drop-off coordinates
missing_dropoff = (
    df['dropoff_longitude'].isna() |
    df['dropoff_latitude'].isna()
)

# 4. Pickup location = (0, 0)
invalid_pickup_zero = (
    (df['pickup_longitude'] == 0) &
    (df['pickup_latitude'] == 0)
)

# 5. Drop-off location = (0, 0)
invalid_dropoff_zero = (
    (df['dropoff_longitude'] == 0) &
    (df['dropoff_latitude'] == 0)
)

# 6. Invalid longitude
invalid_longitude = (
    (df['pickup_longitude'].abs() > 180) |
    (df['dropoff_longitude'].abs() > 180)
)

# 7. Invalid latitude
invalid_latitude = (
    (df['pickup_latitude'].abs() > 90) |
    (df['dropoff_latitude'].abs() > 90)
)

# Combine all removal conditions
remove_mask = (
    invalid_fare |
    invalid_passenger |
    missing_dropoff |
    invalid_pickup_zero |
    invalid_dropoff_zero |
    invalid_longitude |
    invalid_latitude
)

print("\nCleaning rules created successfully.")

print("\nRows marked for removal:")
print(remove_mask.sum())

print("\nRows to keep:")
print((~remove_mask).sum())

print("\nPercentage to remove:")
print(f"{remove_mask.mean() * 100:.4f}%")

print("\nPercentage to keep:")
print(f"{(~remove_mask).mean() * 100:.4f}%")

print("=" * 70)
```

```python
# ============================================================
# STEP 6.3 — CALCULATE TRIP DISTANCE USING HAVERSINE FORMULA
# ============================================================

print("=" * 70)
print("STEP 6.3 — CALCULATING TRIP DISTANCE")
print("=" * 70)

# Convert latitude and longitude from degrees to radians
lat1 = np.radians(df_clean['pickup_latitude'].to_numpy())
lon1 = np.radians(df_clean['pickup_longitude'].to_numpy())

lat2 = np.radians(df_clean['dropoff_latitude'].to_numpy())
lon2 = np.radians(df_clean['dropoff_longitude'].to_numpy())

# Difference between coordinates
dlat = lat2 - lat1
dlon = lon2 - lon1

# Haversine formula
a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
)

c = 2 * np.arcsin(np.sqrt(a))

# Earth's radius in kilometers
earth_radius_km = 6371.0

# Calculate distance
df_clean['trip_distance_km'] = earth_radius_km * c

print("\nTrip distance feature created successfully!")

print("\nDistance statistics:")
print(df_clean['trip_distance_km'].describe())

print("\nFirst 10 trip distances:")
print(df_clean['trip_distance_km'].head(10))

print("\nMissing distances:")
print(df_clean['trip_distance_km'].isna().sum())

print("\nInfinite distances:")
print(np.isinf(df_clean['trip_distance_km']).sum())

print("\nDataset shape:")
print(df_clean.shape)

print("=" * 70)
print("STEP 6.3 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 6.6 — DETECT CLEARLY INVALID COORDINATES
# ============================================================

print("=" * 70)
print("STEP 6.6 — DETECTING CLEARLY INVALID COORDINATES")
print("=" * 70)

# ------------------------------------------------------------
# NYC approximate coordinate ranges
# ------------------------------------------------------------

LON_MIN = -74.3
LON_MAX = -73.6

LAT_MIN = 40.4
LAT_MAX = 41.0

# ------------------------------------------------------------
# Check whether pickup coordinates are inside NYC range
# ------------------------------------------------------------

pickup_valid = (
    df_clean['pickup_longitude'].between(LON_MIN, LON_MAX) &
    df_clean['pickup_latitude'].between(LAT_MIN, LAT_MAX)
)

# ------------------------------------------------------------
# Check whether drop-off coordinates are inside NYC range
# ------------------------------------------------------------

dropoff_valid = (
    df_clean['dropoff_longitude'].between(LON_MIN, LON_MAX) &
    df_clean['dropoff_latitude'].between(LAT_MIN, LAT_MAX)
)

# ------------------------------------------------------------
# Both pickup and drop-off valid
# ------------------------------------------------------------

both_valid = pickup_valid & dropoff_valid

# ------------------------------------------------------------
# Both pickup and drop-off outside
# ------------------------------------------------------------

both_invalid = (~pickup_valid) & (~dropoff_valid)

# ------------------------------------------------------------
# Only pickup invalid
# ------------------------------------------------------------

pickup_only_invalid = (~pickup_valid) & dropoff_valid

# ------------------------------------------------------------
# Only drop-off invalid
# ------------------------------------------------------------

dropoff_only_invalid = pickup_valid & (~dropoff_valid)

print("\n1. PICKUP COORDINATES")
print("Valid pickup coordinates:",
      pickup_valid.sum())

print("Invalid pickup coordinates:",
      (~pickup_valid).sum())

print("\n2. DROP-OFF COORDINATES")
print("Valid drop-off coordinates:",
      dropoff_valid.sum())

print("Invalid drop-off coordinates:",
      (~dropoff_valid).sum())

print("\n3. BOTH COORDINATES VALID")
print("Both valid:",
      both_valid.sum())

print("\n4. BOTH COORDINATES INVALID")
print("Both invalid:",
      both_invalid.sum())

print("\n5. ONLY PICKUP INVALID")
print("Pickup invalid, drop-off valid:",
      pickup_only_invalid.sum())

print("\n6. ONLY DROP-OFF INVALID")
print("Drop-off invalid, pickup valid:",
      dropoff_only_invalid.sum())

# ------------------------------------------------------------
# Percentages
# ------------------------------------------------------------

print("\n7. PERCENTAGES")

print(
    "Both valid:",
    f"{(both_valid.sum() / len(df_clean)) * 100:.4f}%"
)

print(
    "At least one invalid:",
    f"{((~both_valid).sum() / len(df_clean)) * 100:.4f}%"
)

print(
    "Both invalid:",
    f"{(both_invalid.sum() / len(df_clean)) * 100:.4f}%"
)

# ------------------------------------------------------------
# Inspect clearly invalid records
# ------------------------------------------------------------

print("\n8. SAMPLE INVALID RECORDS")

invalid_sample = df_clean.loc[
    ~both_valid,
    [
        'fare_amount',
        'pickup_longitude',
        'pickup_latitude',
        'dropoff_longitude',
        'dropoff_latitude',
        'passenger_count',
        'trip_distance_km'
    ]
].head(30)

display(invalid_sample)

print("=" * 70)
print("STEP 6.6 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 6.8 — VALIDATE POSSIBLE COORDINATE SWAPPING
# ============================================================

print("=" * 70)
print("STEP 6.8 — VALIDATING COORDINATE SWAPPING")
print("=" * 70)

# NYC coordinate ranges
LON_MIN = -74.3
LON_MAX = -73.6

LAT_MIN = 40.4
LAT_MAX = 41.0

# ------------------------------------------------------------
# Pickup: swap longitude and latitude
# ------------------------------------------------------------

swapped_pickup_longitude = df_clean['pickup_latitude']
swapped_pickup_latitude = df_clean['pickup_longitude']

pickup_valid_after_swap = (
    swapped_pickup_longitude.between(LON_MIN, LON_MAX) &
    swapped_pickup_latitude.between(LAT_MIN, LAT_MAX)
)

# ------------------------------------------------------------
# Drop-off: swap longitude and latitude
# ------------------------------------------------------------

swapped_dropoff_longitude = df_clean['dropoff_latitude']
swapped_dropoff_latitude = df_clean['dropoff_longitude']

dropoff_valid_after_swap = (
    swapped_dropoff_longitude.between(LON_MIN, LON_MAX) &
    swapped_dropoff_latitude.between(LAT_MIN, LAT_MAX)
)

# ------------------------------------------------------------
# Rows where the coordinate-order pattern was detected
# ------------------------------------------------------------

pickup_swapped_pattern = (
    df_clean['pickup_longitude'].between(LAT_MIN, LAT_MAX) &
    df_clean['pickup_latitude'].between(LON_MIN, LON_MAX)
)

dropoff_swapped_pattern = (
    df_clean['dropoff_longitude'].between(LAT_MIN, LAT_MAX) &
    df_clean['dropoff_latitude'].between(LON_MIN, LON_MAX)
)

# ------------------------------------------------------------
# How many become valid after swapping?
# ------------------------------------------------------------

pickup_recoverable = (
    pickup_swapped_pattern &
    pickup_valid_after_swap
)

dropoff_recoverable = (
    dropoff_swapped_pattern &
    dropoff_valid_after_swap
)

either_recoverable = (
    pickup_recoverable |
    dropoff_recoverable
)

print("\n1. PICKUP")

print("Possible swapped pickup rows:",
      pickup_swapped_pattern.sum())

print("Pickup rows valid after swapping:",
      pickup_recoverable.sum())

print("\n2. DROP-OFF")

print("Possible swapped drop-off rows:",
      dropoff_swapped_pattern.sum())

print("Drop-off rows valid after swapping:",
      dropoff_recoverable.sum())

print("\n3. EITHER PICKUP OR DROP-OFF")

print("Rows recoverable by swapping:",
      either_recoverable.sum())

print(
    "Percentage:",
    f"{(either_recoverable.sum() / len(df_clean)) * 100:.4f}%"
)

# ------------------------------------------------------------
# Show examples
# ------------------------------------------------------------

print("\n4. SAMPLE RECOVERABLE RECORDS")

recoverable_sample = df_clean.loc[
    either_recoverable,
    [
        'fare_amount',
        'pickup_longitude',
        'pickup_latitude',
        'dropoff_longitude',
        'dropoff_latitude',
        'trip_distance_km'
    ]
].head(20)

display(recoverable_sample)

print("=" * 70)
print("STEP 6.8 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 6.9 — CORRECT SWAPPED COORDINATES
# ============================================================

print("=" * 70)
print("STEP 6.9 — CORRECTING SWAPPED COORDINATES")
print("=" * 70)

# NYC coordinate ranges
LON_MIN = -74.3
LON_MAX = -73.6

LAT_MIN = 40.4
LAT_MAX = 41.0

# ------------------------------------------------------------
# Detect swapped pickup coordinates
# ------------------------------------------------------------

pickup_swapped = (
    df_clean['pickup_longitude'].between(LAT_MIN, LAT_MAX) &
    df_clean['pickup_latitude'].between(LON_MIN, LON_MAX)
)

# ------------------------------------------------------------
# Detect swapped drop-off coordinates
# ------------------------------------------------------------

dropoff_swapped = (
    df_clean['dropoff_longitude'].between(LAT_MIN, LAT_MAX) &
    df_clean['dropoff_latitude'].between(LON_MIN, LON_MAX)
)

print("\nRows with swapped pickup coordinates:",
      pickup_swapped.sum())

print("Rows with swapped drop-off coordinates:",
      dropoff_swapped.sum())

# ------------------------------------------------------------
# Correct pickup coordinates
# ------------------------------------------------------------

temp_pickup_longitude = df_clean.loc[
    pickup_swapped, 'pickup_longitude'
].copy()

df_clean.loc[
    pickup_swapped, 'pickup_longitude'
] = df_clean.loc[
    pickup_swapped, 'pickup_latitude'
].values

df_clean.loc[
    pickup_swapped, 'pickup_latitude'
] = temp_pickup_longitude.values

# ------------------------------------------------------------
# Correct drop-off coordinates
# ------------------------------------------------------------

temp_dropoff_longitude = df_clean.loc[
    dropoff_swapped, 'dropoff_longitude'
].copy()

df_clean.loc[
    dropoff_swapped, 'dropoff_longitude'
] = df_clean.loc[
    dropoff_swapped, 'dropoff_latitude'
].values

df_clean.loc[
    dropoff_swapped, 'dropoff_latitude'
] = temp_dropoff_longitude.values

print("\nCoordinate correction completed.")

# ------------------------------------------------------------
# Verify corrected coordinates
# ------------------------------------------------------------

print("\nVerification:")

remaining_pickup_swapped = (
    df_clean['pickup_longitude'].between(LAT_MIN, LAT_MAX) &
    df_clean['pickup_latitude'].between(LON_MIN, LON_MAX)
).sum()

remaining_dropoff_swapped = (
    df_clean['dropoff_longitude'].between(LAT_MIN, LAT_MAX) &
    df_clean['dropoff_latitude'].between(LON_MIN, LON_MAX)
).sum()

print(
    "Remaining swapped pickup patterns:",
    remaining_pickup_swapped
)

print(
    "Remaining swapped drop-off patterns:",
    remaining_dropoff_swapped
)

print("\nDataset shape:")
print(df_clean.shape)

print("=" * 70)
print("STEP 6.9 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 6.10 — RECALCULATE TRIP DISTANCE
# ============================================================

print("=" * 70)
print("STEP 6.10 — RECALCULATING TRIP DISTANCE")
print("=" * 70)

# Convert coordinates to radians
lat1 = np.radians(df_clean['pickup_latitude'].to_numpy())
lon1 = np.radians(df_clean['pickup_longitude'].to_numpy())

lat2 = np.radians(df_clean['dropoff_latitude'].to_numpy())
lon2 = np.radians(df_clean['dropoff_longitude'].to_numpy())

# Differences
dlat = lat2 - lat1
dlon = lon2 - lon1

# Haversine formula
a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
)

c = 2 * np.arcsin(np.sqrt(a))

# Earth's radius in kilometers
earth_radius_km = 6371.0

# Recalculate distance
df_clean['trip_distance_km'] = earth_radius_km * c

print("\nTrip distance recalculated successfully.")

print("\nDistance statistics:")
print(df_clean['trip_distance_km'].describe())

print("\nMissing distances:")
print(df_clean['trip_distance_km'].isna().sum())

print("\nInfinite distances:")
print(np.isinf(df_clean['trip_distance_km']).sum())

print("\nZero-distance trips:")
print((df_clean['trip_distance_km'] == 0).sum())

print("\nDataset shape:")
print(df_clean.shape)

print("=" * 70)
print("STEP 6.10 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# SECTION 12.4 — IDENTIFY POTENTIALLY REDUNDANT FEATURES
# ============================================================

print("=" * 70)
print("SECTION 12.4 — POTENTIALLY REDUNDANT FEATURES")
print("=" * 70)

# Use the correlation matrix already calculated in Section 12.1
corr = df_clean[numeric_features].corr()

# Consider only the upper triangle of the correlation matrix
# to avoid checking the same pair twice.

upper_triangle = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

# Find highly correlated feature pairs
threshold = 0.80

redundant_pairs = []

for column in upper_triangle.columns:
    for row in upper_triangle.index:
        value = upper_triangle.loc[row, column]

        if pd.notna(value) and abs(value) >= threshold:
            redundant_pairs.append(
                (row, column, value)
            )

# Display results
print(f"\nCorrelation threshold used: |correlation| >= {threshold}")

if redundant_pairs:
    print("\nPotentially redundant feature pairs:")

    redundant_df = pd.DataFrame(
        redundant_pairs,
        columns=[
            'Feature 1',
            'Feature 2',
            'Correlation'
        ]
    )

    redundant_df = redundant_df.sort_values(
        'Correlation',
        key=lambda x: x.abs(),
        ascending=False
    )

    display(redundant_df)

else:
    print("\nNo feature pairs found with |correlation| >= 0.80.")

print("\nImportant:")
print("High correlation indicates potential redundancy,")
print("but features should not be removed automatically.")

print("=" * 70)
print("STEP 12.4 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# SECTION 16 — LOSS FUNCTIONS FOR REGRESSION
# SECTION 16.1 — MSE, MAE AND HUBER LOSS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

print("=" * 70)
print("SECTION 16 — LOSS FUNCTIONS FOR REGRESSION")
print("=" * 70)

# ------------------------------------------------------------
# Actual and predicted values
# ------------------------------------------------------------

actual = np.array([10, 10, 10, 10, 10, 10])
predicted = np.array([9, 11, 12, 8, 10, 30])

error = actual - predicted

# ------------------------------------------------------------
# MSE
# ------------------------------------------------------------

mse = np.mean(error ** 2)

# ------------------------------------------------------------
# MAE
# ------------------------------------------------------------

mae = np.mean(np.abs(error))

# ------------------------------------------------------------
# Huber Loss
# ------------------------------------------------------------

delta = 1.0

absolute_error = np.abs(error)

huber = np.where(
    absolute_error <= delta,
    0.5 * error ** 2,
    delta * (absolute_error - 0.5 * delta)
)

huber_loss = np.mean(huber)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\nExample Errors:")
print(error)

print("\nLoss Values:")
print("-" * 50)
print(f"MSE        : {mse:.4f}")
print(f"MAE        : {mae:.4f}")
print(f"Huber Loss : {huber_loss:.4f}")

print("=" * 70)
print("SECTION 16.1 COMPLETED")
print("=" * 70)
```

```python

import numpy as np
import pandas as pd
import tensorflow as tf
import time
import gc

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam, RMSprop, SGD

# ==========================================
# 1. Use all training and validation data
# ==========================================
X_tune = X_train_scaled
y_tune = y_train

X_val_tune = X_val_scaled
y_val_tune = y_val

print("Training rows:", len(y_tune))
print("Validation rows:", len(y_val_tune))
print("Total rows used:", len(y_tune) + len(y_val_tune))

# ==========================================
# 2. Your requested search options
# ==========================================
hidden_layers_options = [2, 3, 4]
neurons_options = [32, 64, 128, 256]
learning_rate_options = [0.001, 0.0005, 0.01]
batch_size_options = [16, 32, 64, 128]
dropout_options = [0.0, 0.2, 0.3, 0.5]
optimizer_options = ["Adam", "RMSprop", "SGD"]
loss_options = ["Huber", "MAE", "MSE"]

# Runtime controls
N_TRIALS = 3
MAX_EPOCHS = 12
PATIENCE = 2

rng = np.random.default_rng(42)
tf.random.set_seed(42)
results = []

# ==========================================
# 3. Function to build the model
# ==========================================
def create_tuning_model(
    n_features,
    hidden_layers,
    neurons,
    learning_rate,
    batch_size,
    dropout_rate,
    optimizer_name,
    loss_name
):
    model = Sequential()
    model.add(Input(shape=(n_features,)))

    for _ in range(hidden_layers):
        model.add(Dense(neurons, activation="relu"))

        if dropout_rate > 0:
            model.add(Dropout(dropout_rate))

    model.add(Dense(1))

    if optimizer_name == "Adam":
        optimizer = Adam(learning_rate=learning_rate)
    elif optimizer_name == "RMSprop":
        optimizer = RMSprop(learning_rate=learning_rate)
    else:
        optimizer = SGD(learning_rate=learning_rate)

    if loss_name == "Huber":
        loss_function = tf.keras.losses.Huber()
    elif loss_name == "MAE":
        loss_function = tf.keras.losses.MeanAbsoluteError()
    else:
        loss_function = tf.keras.losses.MeanSquaredError()

    model.compile(
        optimizer=optimizer,
        loss=loss_function,
        metrics=[tf.keras.metrics.MeanAbsoluteError(name="mae")]
    )

    return model

# ==========================================
# 4. Run tuning trials
# ==========================================
for trial in range(N_TRIALS):

    print(f"\n{'=' * 45}")
    print(f"TRIAL {trial + 1}/{N_TRIALS}")
    print(f"{'=' * 45}")

    start_time = time.time()

    hidden_layers = int(rng.choice(hidden_layers_options))
    neurons = int(rng.choice(neurons_options))
    learning_rate = float(rng.choice(learning_rate_options))
    batch_size = int(rng.choice(batch_size_options))
    dropout_rate = float(rng.choice(dropout_options))
    optimizer_name = str(rng.choice(optimizer_options))
    loss_name = str(rng.choice(loss_options))

    print("Hidden layers:", hidden_layers)
    print("Neurons:", neurons)
    print("Learning rate:", learning_rate)
    print("Batch size:", batch_size)
    print("Dropout:", dropout_rate)
    print("Optimizer:", optimizer_name)
    print("Loss:", loss_name)

    model = create_tuning_model(
        n_features=X_tune.shape[1],
        hidden_layers=hidden_layers,
        neurons=neurons,
        learning_rate=learning_rate,
        batch_size=batch_size,
        dropout_rate=dropout_rate,
        optimizer_name=optimizer_name,
        loss_name=loss_name
    )

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=PATIENCE,
        restore_best_weights=True
    )

    history = model.fit(
        X_tune,
        y_tune,
        validation_data=(X_val_tune, y_val_tune),
        epochs=MAX_EPOCHS,
        batch_size=batch_size,
        callbacks=[early_stop],
        verbose=1
    )

    best_idx = int(np.argmin(history.history["val_loss"]))

    results.append({
        "Trial": trial + 1,
        "Hidden_Layers": hidden_layers,
        "Neurons": neurons,
        "Learning_Rate": learning_rate,
        "Batch_Size": batch_size,
        "Dropout": dropout_rate,
        "Optimizer": optimizer_name,
        "Loss": loss_name,
        "Best_Epoch": best_idx + 1,
        "Best_Val_Loss": float(history.history["val_loss"][best_idx]),
        "Best_Val_MAE": float(history.history["val_mae"][best_idx]),
        "Training_Time_Min": round((time.time() - start_time) / 60, 2)
    })

    print("Best validation loss:",
          history.history["val_loss"][best_idx])
    print("Validation MAE:",
          history.history["val_mae"][best_idx])
    print("Trial time (minutes):",
          round((time.time() - start_time) / 60, 2))

    del model, history
    tf.keras.backend.clear_session()
    gc.collect()

# ==========================================
# 5. Compare trials
# ==========================================
tuning_results_df = pd.DataFrame(results)

tuning_results_df = tuning_results_df.sort_values(
    by="Best_Val_MAE",
    ascending=True
).reset_index(drop=True)

display(tuning_results_df)

# ==========================================
# 6. Store the best configuration
# ==========================================
best_config = tuning_results_df.iloc[0]

best_hidden_layers = int(best_config["Hidden_Layers"])
best_neurons = int(best_config["Neurons"])
best_learning_rate = float(best_config["Learning_Rate"])
best_batch_size = int(best_config["Batch_Size"])
best_dropout = float(best_config["Dropout"])
best_optimizer = str(best_config["Optimizer"])
best_loss = str(best_config["Loss"])
best_epochs = int(best_config["Best_Epoch"])

print("\nBEST CONFIGURATION")
print("Hidden layers:", best_hidden_layers)
print("Neurons:", best_neurons)
print("Learning rate:", best_learning_rate)
print("Batch size:", best_batch_size)
print("Dropout:", best_dropout)
print("Optimizer:", best_optimizer)
print("Loss:", best_loss)
print("Epochs:", best_epochs)
```

```python
# ============================================================
# SECTION 20.1 — FINAL TUNED MODEL
# ============================================================

import tensorflow as tf
import numpy as np
import time
import gc

print("=" * 70)
print("SECTION 20.1 — BUILD FINAL TUNED MODEL")
print("=" * 70)

# ------------------------------------------------------------
# Use the best configuration from Section 19.7
# ------------------------------------------------------------

print("\nSelected Hyperparameters:")
print("Hidden Layers :", best_hidden_layers)
print("Neurons       :", best_neurons)
print("Learning Rate :", best_learning_rate)
print("Batch Size    :", best_batch_size)
print("Dropout       :", best_dropout)
print("Optimizer     :", best_optimizer)
print("Loss          :", best_loss)
print("Epochs        :", best_epochs)

# ------------------------------------------------------------
# Create final model
# ------------------------------------------------------------

final_model = tf.keras.Sequential()

final_model.add(
    tf.keras.layers.Input(
        shape=(X_train_scaled.shape[1],)
    )
)

# ------------------------------------------------------------
# Hidden layers
# ------------------------------------------------------------

for _ in range(best_hidden_layers):

    final_model.add(
        tf.keras.layers.Dense(
            best_neurons,
            activation="relu"
        )
    )

    if best_dropout > 0:

        final_model.add(
            tf.keras.layers.Dropout(
                best_dropout
            )
        )

# ------------------------------------------------------------
# Output layer
# ------------------------------------------------------------

final_model.add(
    tf.keras.layers.Dense(1)
)

# ------------------------------------------------------------
# Optimizer
# ------------------------------------------------------------

if best_optimizer == "Adam":

    final_optimizer = tf.keras.optimizers.Adam(
        learning_rate=best_learning_rate
    )

elif best_optimizer == "RMSprop":

    final_optimizer = tf.keras.optimizers.RMSprop(
        learning_rate=best_learning_rate
    )

else:

    final_optimizer = tf.keras.optimizers.SGD(
        learning_rate=best_learning_rate
    )

# ------------------------------------------------------------
# Loss
# ------------------------------------------------------------

if best_loss == "MSE":

    final_loss = tf.keras.losses.MeanSquaredError()

elif best_loss == "MAE":

    final_loss = tf.keras.losses.MeanAbsoluteError()

else:

    final_loss = tf.keras.losses.Huber()

# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

final_model.compile(
    optimizer=final_optimizer,
    loss=final_loss,
    metrics=[
        tf.keras.metrics.MeanAbsoluteError(
            name="mae"
        )
    ]
)

# ------------------------------------------------------------
# Model summary
# ------------------------------------------------------------

final_model.summary()
```

```python
# ============================================================
# SECTION 20.2 — TRAIN FINAL TUNED MODEL
# ============================================================

print("=" * 70)
print("SECTION 20.2 — TRAINING FINAL MODEL")
print("=" * 70)

start_time = time.time()

final_history = final_model.fit(
    X_train_scaled,
    y_train,
    validation_data=(
        X_val_scaled,
        y_val
    ),
    epochs=best_epochs,
    batch_size=best_batch_size,
    verbose=1
)

final_training_time = (
    time.time() - start_time
) / 60

print("\n" + "=" * 70)
print("FINAL MODEL TRAINING COMPLETED")
print("=" * 70)

print(
    "Training Time:",
    round(final_training_time, 2),
    "minutes"
)
```

```python
# ============================================================
# SECTION 20.3 — FINAL MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

print("=" * 70)
print("SECTION 20.3 — TEST SET EVALUATION")
print("=" * 70)

# ------------------------------------------------------------
# Predict fare on unseen test data
# ------------------------------------------------------------

y_pred = final_model.predict(
    X_test_scaled,
    batch_size=best_batch_size,
    verbose=1
).flatten()

# ------------------------------------------------------------
# Calculate evaluation metrics
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\nFINAL TEST SET RESULTS")
print("-" * 50)

print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")

print("\n" + "=" * 70)
print("SECTION 20.3 COMPLETED")
print("=" * 70)
```
